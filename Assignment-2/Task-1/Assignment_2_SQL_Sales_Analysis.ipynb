```python
# 1.Load dataset into a SQL database.*****************************************************************  

import pandas as pd
import sqlite3

# Load CSV
df = pd.read_csv("Downloads\\Sample - Superstore.csv", encoding='latin1')

# Create SQLite database
conn = sqlite3.connect("superstore.db")

# Store dataframe as SQL table
df.to_sql("sales", conn, if_exists="replace", index=False)

print("Data Loaded Successfully")
```

```python
# 2. Explore table (schema, sample data).**************************************************************

pd.read_sql("""
PRAGMA table_info(sales);
""", conn)
```

```python
# sample data

pd.read_sql("""
SELECT * FROM sales LIMIT 10;
""", conn)
```

```python
# 3.Apply WHERE filters (region, category, date, sales).***************************************************************
# Filtered data where sales greater than 500:-
querry = """
SELECT *
FROM sales
WHERE Sales > 500;
"""
pd.read_sql(querry, conn)
```

```python
# Filtered data where Region = West:-
querry = """
SELECT * 
FROM sales
WHERE Region = 'West';
"""
pd.read_sql(querry, conn)
```

```python
# Filter data according to the categories, category = Technology

querry = """
SELECT *
FROM sales
WHERE Category = "Technology";
"""
pd.read_sql(querry, conn)
```

```python
#Filtering the data according to the date.
query = """
SELECT *
FROM sales
WHERE [Order Date] >= '2017-01-01';
"""
pd.read_sql(query, conn)
```

```python
#  4.Use GROUP BY for aggregations (sales, quantity, averages).*******************************************************

# Aggregate the data according to the Total Sales by Category.

query = """
SELECT Category,
       SUM(Sales) AS Total_Sales
FROM sales
GROUP BY Category;
"""
pd.read_sql(query, conn)
```

```python
# Aggregate the data according to the Total Quantity by Category.

query = """
SELECT Category,
       SUM(Quantity) AS Total_Quantity
FROM sales
GROUP BY Category;
"""
pd.read_sql(query, conn)
```

```python
# Aggregate the data according to the Average Sales by Region.

query = """
SELECT Region,
       ROUND(AVG(Sales),2) AS Avg_Sales
FROM sales
GROUP BY Region;
"""
pd.read_sql(query, conn)
```

```python
# Aggregate the data according to the Sales and Quantity by Region.

query = """
SELECT Region,
       SUM(Sales) AS Total_Sales,
       SUM(Quantity) AS Total_Quantity
FROM sales
GROUP BY Region;
"""
pd.read_sql(query, conn)
```

```python
# 5.Sort and limit results (top products, top categories).**********************************************************

# Sorting the limiting the data as Top 10 Products by Sales.

query = """
SELECT [Product Name],
       SUM(Sales) AS Total_Sales
FROM sales
GROUP BY [Product Name]
ORDER BY Total_Sales DESC
LIMIT 10;
"""
pd.read_sql(query, conn)
```

```python
# Top Categories by Sales

query = """
SELECT Category,
       SUM(Sales) AS Total_Sales
FROM sales
GROUP BY Category
ORDER BY Total_Sales DESC;
"""
pd.read_sql(query, conn)
```

```python
# Top 5 Customers

query = """
SELECT [Customer Name],
       SUM(Sales) AS Total_Sales
FROM sales
GROUP BY [Customer Name]
ORDER BY Total_Sales DESC
LIMIT 5;
"""
pd.read_sql(query, conn)
```

```python
# 6.Solve use cases (monthly trends, top customers, duplicates).********************************************************

# Monthly Sales Trend
query = """
SELECT [Order Date],
       SUM(Sales) AS Total_Sales
FROM sales
GROUP BY [Order Date];
"""
pd.read_sql(query, conn)
```

```python
# Top 10 Customers
query = """
SELECT [Customer Name],
       SUM(Sales) AS Total_Sales
FROM sales
GROUP BY [Customer Name]
ORDER BY Total_Sales DESC
LIMIT 10;
"""
pd.read_sql(query, conn)
```

```python
# Find Duplicate Records
query = """
SELECT [Order ID],
       COUNT(*) AS Duplicate_Count
FROM sales
GROUP BY [Order ID]
HAVING COUNT(*) > 1;
"""
pd.read_sql(query, conn)
```

```python
# 7.Validate results (row counts, data quality).****************************************************************

# Total Number of Rows
query = """
SELECT COUNT(*) AS Total_Rows
FROM sales;
"""
pd.read_sql(query, conn)
```

```python
# Count Null Sales Values
query = """
SELECT COUNT(*) AS Null_Sales
FROM sales
WHERE Sales IS NULL;
"""
pd.read_sql(query, conn)
```

```python
# Count Distinct Customers

query = """
SELECT COUNT(DISTINCT [Customer Name]) AS Total_Customers
FROM sales;
"""
pd.read_sql(query, conn)
```

```python
# Count Distinct Products
query = """
SELECT COUNT(DISTINCT [Product Name]) AS Total_Products
FROM sales;
"""
pd.read_sql(query, conn)
```
