##Query 1 — Customer Overview

```python
%sql

SELECT *
FROM workspace.default.gold_customer360
ORDER BY total_spent DESC
LIMIT 10;
```

##Query 2 — Monthly Revenue Trend

```python

%sql

SELECT *
FROM workspace.default.gold_monthly_sales
ORDER BY sales_month;
```

##Query 3 — Top Revenue Products

```python
%sql

SELECT *
FROM workspace.default.gold_product_performance
ORDER BY total_revenue DESC
LIMIT 10;
```

##Query 4 — Most Popular Payment Method

```python
%sql

SELECT *
FROM workspace.default.gold_payment_summary
ORDER BY number_of_payments DESC;
```

##Query 5 — Delivery Performance

```python
%sql

SELECT *
FROM workspace.default.gold_delivery_performance
ORDER BY avg_delivery_days;
```

##Query 6 — Total Revenue

```python
%sql

SELECT
SUM(total_revenue) AS company_revenue
FROM workspace.default.gold_monthly_sales;
```

##Query 7 — Total Customers

```python
%sql

SELECT
COUNT(*) AS total_customers
FROM workspace.default.gold_customer360;
```

##Query 8 — Average Order Value

```python
%sql

SELECT
ROUND(AVG(avg_order_value),2) AS average_order_value
FROM workspace.default.gold_customer360;
```

##Query 9 — Best Selling Product Category

```python
%sql

SELECT
product_category_name,
SUM(units_sold) AS units
FROM workspace.default.gold_product_performance
GROUP BY product_category_name
ORDER BY units DESC
LIMIT 10;
```

##Query 10 — Highest Revenue MonthQuery

```python
%sql

SELECT *
FROM workspace.default.gold_monthly_sales
ORDER BY total_revenue DESC
LIMIT 1;
```

## Business Impact

The SQL analytics layer enables RetailMart stakeholders to answer key business questions without querying raw transactional data.

Using Gold tables reduces query complexity, improves performance, and provides a single source of truth for dashboards and reports.
