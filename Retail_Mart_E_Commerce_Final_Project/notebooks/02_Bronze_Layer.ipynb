# Bronze Layer - Raw Data Ingestion

### Read All Raw Files

```python
customers_df = (
    spark.read
         .option("header", True)
         .option("inferSchema", True)
         .csv("/Volumes/workspace/default/retailmart_raw/raw_customers_dataset.csv")
)

orders_df = (
    spark.read
         .option("header", True)
         .option("inferSchema", True)
         .csv("/Volumes/workspace/default/retailmart_raw/raw_orders_dataset.csv")
)

items_df = (
    spark.read
         .option("header", True)
         .option("inferSchema", True)
         .csv("/Volumes/workspace/default/retailmart_raw/raw_items_dataset.csv")
)

payments_df = (
    spark.read
         .option("header", True)
         .option("inferSchema", True)
         .csv("/Volumes/workspace/default/retailmart_raw/raw_payments_dataset.csv")
)

products_df = (
    spark.read
         .option("header", True)
         .option("inferSchema", True)
         .csv("/Volumes/workspace/default/retailmart_raw/raw_products_dataset.csv")
)
```

```python
print("Customers :", customers_df.count())
print("Orders    :", orders_df.count())
print("Items     :", items_df.count())
print("Payments  :", payments_df.count())
print("Products  :", products_df.count())
```

```python
customers_df.printSchema()
orders_df.printSchema()
items_df.printSchema()
payments_df.printSchema()
products_df.printSchema()
```

```python
customers_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.bronze_customers")

orders_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.bronze_orders")

items_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.bronze_items")

payments_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.bronze_payments")

products_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.bronze_products")
```
